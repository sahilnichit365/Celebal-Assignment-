```python
spark.sql("SELECT current_catalog(), current_schema()").show()
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS lms_analytics")
```

```python
spark.sql("USE lms_analytics")
```

```python
learners_df.write.format("delta").mode("overwrite").saveAsTable("lms_analytics.bronze_learners")
```

```python
courses_df.write.format("delta").mode("overwrite").saveAsTable("lms_analytics.bronze_courses")
```

```python
enrolment_activity_df.write.format("delta").mode("overwrite").saveAsTable("lms_analytics.bronze_enrolment_activity")
```
