Step 4: Data Transformation:
Create Age column from Year_Birth

```python
from pyspark.sql.functions import year, current_date, col

df = df.withColumn(
    "Age",
    year(current_date()) - col("Year_Birth")
)

df.select("Year_Birth", "Age").show(5)
```

```python
from pyspark.sql.functions import when

df = df.withColumn(
    "Region",
    when((col("ID") % 4) == 0, "North")
    .when((col("ID") % 4) == 1, "South")
    .when((col("ID") % 4) == 2, "East")
    .otherwise("West")
)

df.select("ID", "Region").show(10)
```

```python
from pyspark.sql.functions import when

df = df.withColumn(
    "Category",
    when(col("MntWines") > 500, "Premium")
    .otherwise("Regular")
)

df.select("MntWines", "Category").show(10)
```

## Step 4: Data Cleaning
- Checking duplicate records
- Checking missing values
- Handling null values
- Identifying inconsistent data

```python
print("Rows before removing duplicates:", df.count())
df = df.dropDuplicates()
print("Rows after removing duplicates:", df.count())
```

```python
from pyspark.sql.functions import col, sum
null_counts = df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
])

null_counts.show(truncate=False)
```

```python
from pyspark.sql.functions import avg

average_income = df.select(avg("Income")).first()[0]
df = df.na.fill({"Income": average_income})
df.select("Income").show(10)
```

```python
df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```

## Step 5: Data Transformation

- Renaming columns for better readability.
- Converting the customer joining date to DateType.
- Creating a new Age column using the Year_Birth column.

```python
df = df.withColumnRenamed(
    "Dt_Customer",
    "Customer_Join_Date"
)

print(df.columns)
```

```python
from pyspark.sql.functions import to_date, col

df = df.withColumn(
    "Customer_Join_Date",
    to_date(col("Customer_Join_Date"), "dd-MM-yyyy")
)

df.printSchema()
```

## Step 7: Data Aggregation

Aggregation is used to summarize data by performing calculations such as:
- Count()
- Sum()
- Average()
- Minimum()
- Maximum()

```python
print("Total Customers :", df.count())
```

```python
from pyspark.sql.functions import sum

df.select(
    sum("MntWines").alias("Total_Wine_Spending")
).show()
```

```python
from pyspark.sql.functions import avg

df.select(
    avg("Income").alias("Average_Income")
).show()
```

## Step 8: GroupBy Operations

Grouping allows us to perform calculations for each category separately.

```python
df.groupBy("Education").count().show()
```

```python
df.groupBy("Marital_Status").agg(
      avg("MntWines").alias("Average_Wine_Spending")
  ).show()
```

```python
from pyspark.sql.functions import avg, min, max

df.agg(
    min("Income").alias("Minimum_Income"),
    max("Income").alias("Maximum_Income"),
    avg("Income").alias("Average_Income")
).show()
```

```python
df.groupBy("Education").agg(
      avg("Income").alias("Average_Income"),
      max("Income").alias("Maximum_Income"),
      min("Income").alias("Minimum_Income"),
      count("*").alias("Total_Customers")
  ).show()
```

## Step 10: Complete Data Processing Pipeline

This pipeline demonstrates a complete Spark workflow:
1. Remove duplicate records.
2. Fill missing Income values.
3. Create Age column.
4. Filter customers with Income greater than 50,000.
5. Group data by Education.
6. Calculate average income and customer count.

```python
from pyspark.sql.functions import avg, count, col

df = df.dropDuplicates()

average_income = df.select(avg("Income")).first()[0]
df = df.na.fill({"Income": average_income})

from pyspark.sql.functions import year, current_date

df = df.withColumn(
    "Age",
    year(current_date()) - col("Year_Birth")
)

filtered_df = df.filter(col("Income") > 50000)

result = filtered_df.groupBy("Education").agg(
    avg("Income").alias("Average_Income"),
    count("*").alias("Customer_Count")
)

result.show()
```

```python
result.write.mode("overwrite").csv("output/results.csv", header=True)
```
