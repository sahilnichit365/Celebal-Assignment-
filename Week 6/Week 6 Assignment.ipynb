Handle NULL Values

```python
from pyspark.sql.functions import col, count, when

df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()
```

```python
clean_df = df.dropna(subset=["Income"])

clean_df.show(5)
```

```python
print("Original Rows :", df.count())
print("After Removing Nulls :", clean_df.count())
```

```python
renamed_df = clean_df.withColumnRenamed(
    "MntWines",
    "Wine_Spending"
)

renamed_df.show(5)
```

Change Data Type

```python
cast_df = renamed_df.withColumn(
    "Income",
    col("Income").cast("double")
)

cast_df.printSchema()
```

Calculate Total Spending

```python
from pyspark.sql.functions import col

df_total = clean_df.withColumn(
    "Total_Spending",
    col("MntWines") +
    col("MntFruits") +
    col("MntMeatProducts") +
    col("MntFishProducts") +
    col("MntSweetProducts") +
    col("MntGoldProds")
)

df_total.select(
    "ID",
    "Total_Spending"
).show(10)
```

Adding Age Column

```python
from pyspark.sql.functions import lit

df_total = df_total.withColumn(
    "Age",
    lit(2026) - col("Year_Birth")
)

df_total.select(
    "Year_Birth",
    "Age"
).show(10)
```

```python
from pyspark.sql.functions import when

df_total = df_total.withColumn(
    "Customer_Type",
    when(col("Income") >= 70000, "High Income")
    .when(col("Income") >= 40000, "Medium Income")
    .otherwise("Low Income")
)

df_total.select(
    "Income",
    "Customer_Type"
).show(10)
```

## Transformations and Actions

Transformations create a new DataFrame but are not executed immediately because Spark uses Lazy Evaluation. Actions trigger the execution of all pending transformations.

```python
filtered_df = df_total.filter(col("Income") > 50000)
```

Action

```python
filtered_df.show(5)
```

##Lazy Evaluation

```python
lazy_df = (
    df_total
    .filter(col("Income") > 50000)
    .select("Education", "Income", "Total_Spending")
    .orderBy(col("Income").desc())
)
```

```python
lazy_df.show()
```

spark creates a DAG (direct acyclic graph) and executes all transformations together.

## Data Pipeline
Read → Transform → Filter → Write

```python
pipeline_df = (
    clean_df
    .withColumn(
        "Total_Spending",
        col("MntWines")
        + col("MntFruits")
        + col("MntMeatProducts")
        + col("MntFishProducts")
        + col("MntSweetProducts")
        + col("MntGoldProds")
    ).filter(col("Income") > 50000)
    .select(
        "ID",
        "Education",
        "Income",
        "Total_Spending"
    )
)

pipeline_df.show(10)
```

Save As CSV


```python
pipeline_df.write \
.mode("overwrite") \
.option("header", True) \
.csv("output_csv")
```

Save as Parquet

```python
pipeline_df.write \
.mode("overwrite") \
.parquet("output_parquet")
```

Read Parquet

```python
parquet_df = spark.read.parquet("output_parquet")

parquet_df.show(5)
```
